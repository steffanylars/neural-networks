##La regla de aprendizaje del Perceptrón

```python
# Código para creación y prueba de un perceptrón
import numpy, random, os

# calcula la salida del perceptrón para un caso
def predict(x, y):
    outputP = x*weights[0] + y*weights[1] + bias*weights[2]
    if outputP > 0 : #función de activación escalón
       outputP = 1
    else :
       outputP = 0
    return outputP

# despliega la salida del perceptrón
def test(x, y, op):
    print(x, op, y, "is : ", predict(x, y))

# ajusta los pesos del perceptrón con un ejemplo
def perceptron(input1, input2, output) :
   outputP = predict(input1, input2)
   error = output - outputP
   weights[0] += error * input1 * lr
   weights[1] += error * input2 * lr
   weights[2] += error * bias * lr

lr = 0.01 #razón de aprendizaje
bias = 1 #valor del sesgo
```

```python
# Se entrena el perceptrón para el OR
print("Perceptrón para operador OR - linealmente separable")

weights = [random.random(),random.random(),random.random()]
#3 pesos generados al azar para 2 neuronas y el bias)

print("Pesos antes de entrenar el perceptrón")
print(weights)
# evalua los casos
test(1,1,"OR")
test(1,0,"OR")
test(0,1,"OR")
test(0,0,"OR")

#ejecuta 50 epochs de entrenamiento para el OR
for i in range(50) :
    perceptron(1,1,1) 
    perceptron(1,0,1) 
    perceptron(0,1,1) 
    perceptron(0,0,0) 

print("Pesos después de entrenar el perceptrón")
print(weights)

# evalua los casos
test(1,1,"OR")
test(1,0,"OR")
test(0,1,"OR")
test(0,0,"OR")
```

```python
# Se entrena el perceptrón para el AND
print("Perceptrón para operador AND - linealmente separable")

weights = [random.random(),random.random(),random.random()]
#weights generated in a list (3 weights in total for 2 neurons and the bias)

print("Pesos antes de entrenar el perceptrón")
print(weights)
# evalua los casos
test(1,1,"AND")
test(1,0,"AND")
test(0,1,"AND")
test(0,0,"AND")

#ejecuta 50 epochs de entrenamiento para el AND
for i in range(50) :
    perceptron(1,1,1) 
    perceptron(1,0,0) 
    perceptron(0,1,0) 
    perceptron(0,0,0) 

print("Pesos después de entrenar el perceptrón")
print(weights)

# evalua los casos
test(1,1,"AND")
test(1,0,"AND")
test(0,1,"AND")
test(0,0,"AND")
```

```python
# Se entrena el perceptrón para el XOR
print("Perceptrón para operador XOR - no linealmente separable")

weights = [random.random(),random.random(),random.random()]
#weights generated in a list (3 weights in total for 2 neurons and the bias)

print("Pesos antes de entrenar el perceptrón")
print(weights)
# evalua los casos
test(1,1,"XOR")
test(1,0,"XOR")
test(0,1,"XOR")
test(0,0,"XOR")

#ejecuta 50 epochs de entrenamiento para el XOR
for i in range(50) :
    perceptron(1,1,0) 
    perceptron(1,0,1) 
    perceptron(0,1,1) 
    perceptron(0,0,0) 

print("Pesos después de entrenar el perceptrón")
print(weights)

# evalua los casos
test(1,1,"XOR")
test(1,0,"XOR")
test(0,1,"XOR")
test(0,0,"XOR")
```
